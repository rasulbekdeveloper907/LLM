```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("gpt2")

sentences = [
    "I love Machine Learning",
    "Python is amazing",
    "Artificial Intelligence",
    "Hello, how are you?",
    "I am learning NLP"
]

for sentence in sentences:
    print("=" * 50)
    print("Sentence:", sentence)
    print("Tokens:", tokenizer.tokenize(sentence))
    print("Token count:", len(tokenizer.tokenize(sentence)))
```

```python
sentences = [
    "Machine Learning is powerful",
    "Natural Language Processing is interesting",
    "I am learning Artificial Intelligence",
    "ChatGPT can understand human language"
]

for sentence in sentences:
    words = sentence.split()
    tokens = tokenizer.tokenize(sentence)

    print("=" * 50)
    print("Sentence:", sentence)
    print("Words:", words)
    print("Word count:", len(words))
    print("Tokens:", tokens)
    print("Token count:", len(tokens))
```

```python
words = [
    "MachineLearning",
    "DeepLearning",
    "ArtificialIntelligence",
    "NaturalLanguageProcessing",
    "Transformers"
]

for word in words:
    tokens = tokenizer.tokenize(word)

    print("=" * 40)
    print("Word:", word)
    print("Tokens:", tokens)
    print("Number of tokens:", len(tokens))
```

```python
sentences = [
    "Hello world",
    "Machine Learning",
    "Python",
    "ChatGPT"
]

for sentence in sentences:
    tokens = tokenizer.tokenize(sentence)
    ids = tokenizer.encode(sentence, add_special_tokens=False)

    print("=" * 50)
    print("Sentence:", sentence)
    print("Tokens:", tokens)
    print("Token IDs:", ids)
```

```python
sentence = "Machine Learning is interesting"

tokens = tokenizer.tokenize(sentence)
ids = tokenizer.encode(sentence, add_special_tokens=False)
decoded = tokenizer.decode(ids)

print("Original :", sentence)
print("Tokens   :", tokens)
print("IDs      :", ids)
print("Decoded  :", decoded)
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("gpt2")

texts = [
    "ChatGPT is an AI model.",
    "Machine Learning is interesting.",
    "I am learning NLP.",
    "Assalomu alaykum, bugun NLP o'rganamiz!",
    "ArtificialIntelligence",
    "Deep Learning"
]

for text in texts:

    tokens = tokenizer.tokenize(text)
    ids = tokenizer.encode(text, add_special_tokens=False)

    print("=" * 60)
    print("TEXT       :", text)
    print("WORDS      :", len(text.split()))
    print("TOKENS     :", len(tokens))
    print("TOKEN LIST :", tokens)
    print("TOKEN IDS  :", ids)
    print("RATIO      :", round(len(tokens) / len(text.split()), 2))
```
