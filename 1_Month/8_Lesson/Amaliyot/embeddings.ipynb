```python
from transformers import AutoTokenizer, AutoModel

tokenizer = AutoTokenizer.from_pretrained("gpt2")
model = AutoModel.from_pretrained("gpt2")

print(model)
```

```python
print("Vocabulary size:", model.config.vocab_size)
print("Embedding dimension:", model.config.n_embd)
```

```python
token_id = 40

embedding = model.wte.weight[token_id]

print("Token ID:", token_id)
print("Embedding shape:", embedding.shape)
print("First 10 values:", embedding[:10])
```

```python
word = "AI"

ids = tokenizer.encode(word, add_special_tokens=False)

print("Word:", word)
print("Token IDs:", ids)

for token_id in ids:
    embedding = model.wte.weight[token_id]

    print("Token ID:", token_id)
    print("Embedding shape:", embedding.shape)
    print("First 5 values:", embedding[:5])
```

```python
words = ["cat", "dog", "car", "computer", "apple"]

for word in words:

    ids = tokenizer.encode(word, add_special_tokens=False)

    print("=" * 40)
    print("Word:", word)
    print("IDs:", ids)

    for token_id in ids:
        embedding = model.wte.weight[token_id]

        print("Embedding shape:", embedding.shape)
        print("First 5:", embedding[:5])
```

```python
text = "I love AI"

inputs = tokenizer(
    text,
    return_tensors="pt",
    add_special_tokens=False
)

print("Input IDs:", inputs["input_ids"])

embeddings = model.wte(inputs["input_ids"])

print("Embedding shape:", embeddings.shape)
```

```python
word1 = "cat"
word2 = "dog"

id1 = tokenizer.encode(word1, add_special_tokens=False)[0]
id2 = tokenizer.encode(word2, add_special_tokens=False)[0]

emb1 = model.wte.weight[id1]
emb2 = model.wte.weight[id2]

print("Cat embedding:", emb1[:10])
print("Dog embedding:", emb2[:10])
```

```python
import torch
import torch.nn.functional as F

words = ["cat", "dog", "car"]

embeddings = {}

for word in words:
    token_id = tokenizer.encode(
        word,
        add_special_tokens=False
    )[0]

    embeddings[word] = model.wte.weight[token_id]

similarity = F.cosine_similarity(
    embeddings["cat"].unsqueeze(0),
    embeddings["dog"].unsqueeze(0)
)

print("cat vs dog:", similarity.item())
```

```python
similarity = F.cosine_similarity(
    embeddings["cat"].unsqueeze(0),
    embeddings["car"].unsqueeze(0)
)

print("cat vs car:", similarity.item())
```

```python
import torch
import torch.nn.functional as F

words = ["cat", "dog", "car", "apple"]

embeddings = {}

for word in words:

    token_id = tokenizer.encode(
        word,
        add_special_tokens=False
    )[0]

    embeddings[word] = model.wte.weight[token_id]


for word1 in words:
    for word2 in words:

        similarity = F.cosine_similarity(
            embeddings[word1].unsqueeze(0),
            embeddings[word2].unsqueeze(0)
        )

        print(
            f"{word1:>6} vs {word2:<6} = "
            f"{similarity.item():.4f}"
        )
```

```python
text = "I love Machine Learning"

inputs = tokenizer(
    text,
    return_tensors="pt",
    add_special_tokens=False
)

embeddings = model.wte(inputs["input_ids"])

sentence_embedding = embeddings.mean(dim=1)

print("Token embeddings:", embeddings.shape)
print("Sentence embedding:", sentence_embedding.shape)
```
